```python
import  numpy as np
```

```python
import pandas as pd
```

```python
from sklearn.datasets import load_iris
```

```python
x=pd.DataFrame(load_iris().data,columns=load_iris().feature_names)
```

```python
x
```

```python
y=pd.DataFrame(load_iris().target,columns=['target'])
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
!pip install xgboost
```

```python
from xgboost import XGBClassifier
```

```python
model=XGBClassifier(n_estimators=100,max_depth=6,subsample=0.8,colsample_bytree=0.8,booster='dart',eval_metric='rmse',random_state=42,objective='multi:softprob',reg_alpha=1.0)
```

```python
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
```

```python
accuracy_score(y_test,y_pred)
```

```python
confusion_matrix(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```
